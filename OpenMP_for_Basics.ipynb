```python
%%bash
cat <<EOF > hello_parallel.cpp
#include <iostream>

// include library to use OpenMP
#include <omp.h>

using namespace std;
int main() {

#pragma omp parallel
{
cout << "Hello, World \n" << endl;
cout<<"Hello PDC form parallel region"<< endl;
}

cout << "Hello, World form outside world \n" << endl;

return 0;
}
EOF
```

```python
%%bash
g++ -fopenmp hello_parallel.cpp -o output_hello_parallel
```

```python
%%bash
./output_hello_parallel
```

```python
%%bash
cat <<EOF > hello_N_threads.cpp
#include <iostream>

// include library to use OpenMP
#include <omp.h>

using namespace std;
int main() {

omp_set_num_threads(10);
#pragma omp parallel
{
cout << "Hello, World \n" <<omp_get_thread_num()<< endl;
 // cout<<"Hello PDC form parallel region"<< endl;
}

cout << "Hello, World form outside world \n" << endl;

return 0;
}
EOF
```

```python
%%bash
g++ -fopenmp hello_N_threads.cpp -o hello_N_threads
./hello_N_threads
```

```python
%%bash
cat <<EOF > hello_critical.cpp
#include <iostream>

// include library to use OpenMP
#include <omp.h>

using namespace std;
int main() {

omp_set_num_threads(10);
#pragma omp parallel
{
#pragma omp critical
{
cout << "Hello, World \n" <<omp_get_thread_num()<< endl;
 }
}

cout << "Hello, World form outside world \n" << endl;

return 0;
}
EOF
```

```python
%%bash
g++ -fopenmp hello_critical.cpp -o hello_c
./hello_c
```

```python
%%bash
cat <<EOF > hello_single.cpp
#include <iostream>

// include library to use OpenMP
#include <omp.h>

using namespace std;
int main() {

omp_set_num_threads(10);
#pragma omp parallel
{
#pragma omp single
{
cout << "Hello, World \n" <<omp_get_thread_num()<< endl;
 }
}

cout << "Hello, World form outside world \n" << endl;

return 0;
}
EOF
```

```python
%%bash
g++ -fopenmp hello_single.cpp -o single_c
./single_c
```
